```python

import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
sns.set_style('whitegrid')
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
titanic.sample(15)
```

```python
titanic.info()
```

```python
titanic['embarked'] = titanic['embarked'].astype('category')
```

```python
titanic['embark_town'] = titanic['embark_town'].astype('category')
```

```python
titanic.isna().sum()
```

```python
titanic['age'] = titanic['age'].fillna(titanic['age'].mean())
```

```python
titanic['embarked'] = titanic['embarked'].fillna(titanic['embarked'].mode()[0])
```

```python
titanic['embark_town'] = titanic['embark_town'].fillna(titanic['embark_town'].mode()[0])
```

```python
titanic.head()
```

```python
sns.jointplot(data = titanic, x = 'fare',y = 'age')
plt.show()
```

```python
plt.hist(data = titanic,x = 'fare',bins= 50,color = '#FFE5B4')
plt.ylim(0,500)
plt.show()
```

```python
sns.boxplot(data= titanic,x = 'class',y = 'age',palette=["blue", "green", "orange"])
plt.show()
```

```python
sns.swarmplot(data= titanic,x = 'class',y = 'age',palette=["blue", "green", "orange"])
plt.show()
```

```python
total = titanic.shape[0]
sns.countplot(data=titanic, x='sex',palette=["blue", "green"])
```

```python
n_df = titanic.select_dtypes(include=['number','bool'])
```

```python
corr_matrix = n_df.corr()
```

```python
plt.title('titanic.corr()')
sns.heatmap(corr_matrix,annot = True,cmap = 'coolwarm')
plt.show()
```

```python

sns.displot(data=titanic, x='age', kind='hist',col='sex',bins = 10,alpha = 1)
plt.show()
```
